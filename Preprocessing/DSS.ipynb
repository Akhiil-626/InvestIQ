```python
import pandas as pd

df = pd.read_csv("E:/InvestIQ/Preprocessing/data/preprocessed_with_predictions.csv")
df.head()
```

```python

threshold = 0.5  # if predicted change > $0.5, take action

def generate_signal(row):
    diff = row['Predicted_Close'] - row['Close/Last']
    if diff > threshold:
        return 'Buy'
    elif diff < -threshold:
        return 'Sell'
    else:
        return 'Hold'

df['Decision'] = df.apply(generate_signal, axis=1)
df.to_csv("dss_decisions.csv", index=False)

print(df[['Date', 'Close/Last', 'Predicted_Close', 'Decision']].head())
```
